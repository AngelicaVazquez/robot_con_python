```python
import time

print("=== CONTROL DE PASEO AUTÓNOMO PARA PERROS ===")

# -------------------------------------------------------------
# 1. ENTRADA DE DATOS (INPUTS)
# -------------------------------------------------------------
# Pedimos el nivel de batería inicial
bateria = int(input("¿Qué porcentaje de batería tiene el robot? (0-100): "))

# Preguntamos por la preparación inicial
respuesta_correa = input("¿Le colocaste la correa al perro? (si/no): ").lower()
respuesta_perro = input("¿El perro está listo para salir? (si/no): ").lower()

# Validación de respuestas afirmativas
correa_conectada = (respuesta_correa == "si")
perro_listo = (respuesta_perro == "si")

# -------------------------------------------------------------
# 2. VERIFICACIÓN Y MANEJO DE ERRORES AL INICIAR
# -------------------------------------------------------------
print("\n--- Verificando estado inicial ---")

# Si la correa no está sujeta, el perro no está listo o la batería es menor a 30%
if not correa_conectada or not perro_listo or bateria < 30:
    print("❌ NO SE PUEDE INICIAR EL PASEO:")
    if not correa_conectada:
        print("   - Error: La correa no está colocada correctamente.")
    if not perro_listo:
        print("   - Error: El perro no está preparado.")
    if bateria < 30:
        print("   - Error: La batería es menor a 30% (" + str(bateria) + "%).")
    print("\nEl paseo ha sido cancelado por seguridad.")

else:
    print("✅ Todo correcto. Iniciando el paseo...")

    # Variables de control del paseo
    minuto = 0
    tiempo_maximo = 30
    perro_suelto = False

    # -------------------------------------------------------------
    # 3. BUCLE PRINCIPAL DEL PASEO (mientras minuto < 30)
    # -------------------------------------------------------------
    while minuto < tiempo_maximo:
        minuto = minuto + 5
        bateria = bateria - 5
        print("\n--------------------------------------------------")
        print("📍 Minuto " + str(minuto) + " de " + str(tiempo_maximo) + " | Batería: " + str(bateria) + "%")

        # Regla: Si la batería baja de 30% -> Regresar inmediatamente
        if bateria < 30:
            print("⚠️ Batería baja de 30%. Regresando inmediatamente a casa.")
            break

        # Regla: Si comienza a llover -> Regresar a casa
        lluvia = input("¿Comenzó a llover? (si/no): ").lower()
        if lluvia == "si":
            print("🌧️ Comenzó a llover. Cancelando paseo y regresando a casa.")
            break

        # Regla: Si se desvía del recorrido -> Retornar a la ruta
        desvio = input("¿El robot o perro se desviaron del recorrido? (si/no): ").lower()
        if desvio == "si":
            print("🔄 Recalculando... Retornando a la ruta programada.")

        # Regla: Si detecta vehículos aproximándose -> Detenerse
        vehiculo = input("¿Se aproxima un vehículo? (si/no): ").lower()
        if vehiculo == "si":
            print("🛑 Vehículo detectado. Deteniéndose hasta que pase...")
            time.sleep(1)

        # Regla: Si el perro hace sus necesidades -> Recogerlas
        necesidades = input("¿El perro hizo sus necesidades? (si/no): ").lower()
        if necesidades == "si":
            print("🧹 Recogiendo las necesidades del perro...")

        # Regla: Si encuentra un amigo -> Detenerse a dejarlo jugar
        amigo = input("¿El perro encontró a un amigo? (si/no): ").lower()
        if amigo == "si":
            print("🐕 Amigo detectado. Deteniéndose para dejarlo jugar un momento...")
            time.sleep(1)

        # Regla: Si la zona es segura -> Soltar al perrito para jugar
        zona_segura = input("¿La zona actual es segura (ej. parque cerrado)? (si/no): ").lower()
        if zona_segura == "si":
            perro_suelto = True
            print("🌳 Zona segura confirmada. Soltando al perro de la correa para jugar.")

            # Regla: Mientras el perrito traiga la pelota -> Lanzarla
            trae_pelota = input("¿El perro trae la pelota en el hocico? (si/no): ").lower()
            while trae_pelota == "si":
                print("🎾 Lanzando la pelota...")
                time.sleep(1)
                trae_pelota = input("¿El perro la volvió a traer? (si/no): ").lower()

            # Volver a colocar correa antes de continuar caminando
            print("🔗 Colocando la correa nuevamente para continuar el recorrido.")
            perro_suelto = False

        time.sleep(1) # Pausa simulada entre tramos del paseo

    # -------------------------------------------------------------
    # 4. FINALIZAR Y REGRESAR A CASA
    # -------------------------------------------------------------
    print("\n--------------------------------------------------")
    print("--- FIN DEL PASEO ---")
    print("• El robot y el perro han llegado de regreso a casa.")
    print("• Robot detenido.")

    if perro_suelto:
        print("• Se vuelve a enganchar al perro antes de entrar.")

    correa_conectada = False
    print("• Correa desconectada del perro.")
    print("✅ Proceso completado. Batería final: " + str(bateria) + "%")
```
